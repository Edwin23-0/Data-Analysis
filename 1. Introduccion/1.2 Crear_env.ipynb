## Creación del entorno virtual


Como estamos trabajando en la carpeta base Data Analysis, primero ingresamos a esta ubicación mediante PowerShell. Una vez dentro de la carpeta, ejecutamos el siguiente comando para crear el entorno virtual del proyecto:

```python
python -m venv .venv
```

Esto crea:

Analisis_Datos/

└── .venv/

Luego se activa el entorno:

```python
.\.venv\Scripts\Activate.ps1
```

Si todo está correcto, en la terminal deberías ver algo parecido a:

```python
(.venv) PS C:\...\Analisis_Datos>
```

Ahora se instala las librerías dentro de ese entorno dependiendo del proyecto:

```python
python -m pip install pandas numpy matplotlib seaborn jupyter openpyxl
```

Finalmente, las librerías necesarias ya se encuentran instaladas dentro del entorno virtual. De esta manera, estarán disponibles para ser utilizadas en las diferentes subcarpetas, archivos y notebooks que hagan parte del proyecto de análisis de datos siempre y cuando este entorno se encuentre activado y correctamente configurado.


## Seleccionar Kernel


Como estamos utilizando Jupyter Notebook desde VS Code, es necesario seleccionar el kernel que se utilizará en cada notebook. Por esta razón, debemos instalar ipykernel dentro del entorno virtual, lo que permitirá que Python lo reconozca como un kernel disponible para trabajar en VS Code.

Dentro del mismo entorno activado descargamos:

```python
python -m pip install ipykernel
```

```python
python -m ipykernel install --user --name=analisis-datos --display-name "Python (Analisis Datos)"
```

Después, en VS Code, arriba a la derecha de tu notebook seleccionas:

Kernel → Python (Analisis Datos)
