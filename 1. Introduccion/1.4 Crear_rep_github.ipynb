# Creacion de un repositorio en GITHUB 


Antes de ejecutar comandos se debe tener claro el concepto.

- VS Code: Entorno de trabajo.
- Git: Es el sistema que registra las versiones del proyecto.
- GitHub: Es donde puedes guardar una copia remota del repositorio Git.

## Configuracion GIT

Se descarga git por medio de la web o ejecutando el comando en consola , cuando se descargue se verifica la verison usada por medio de powershell

```python
git --version
```

## Configurar tu identidad de Git

Esto se hace normalmente una sola vez en el computador.


Configura tu nombre:


```python
git config --global user.name "Tu Nombre"
```

Configura tu correo:

```python
git config --global user.email "tu-correo@example.com"
```

Puedes comprobarlo:

```python
git config --global --list
```

## Crear el repositorio Git local

Ahora estás ubicado en:
Data Analysis/

Ejecutar para que Git cree una carpeta oculta:

```python
git init
```

La carpeta .git no debe modificarse manualmente, ya que Git la utiliza para almacenar toda la información y el historial de cambios del proyecto. Por otro lado, el entorno virtual .venv no debe subirse a GitHub porque contiene físicamente todas las librerías instaladas y puede ocupar una gran cantidad de espacio. En su lugar, se utiliza el archivo requirements.txt, que registra las librerías y versiones específicas que necesita el proyecto, permitiendo reconstruir el entorno de trabajo de forma consistente cuando sea necesario.

## Crear correctamente .gitignore

Crear el archivo .gitignore en la carpeta raíz permite indicar a Git qué archivos y carpetas no deben subirse al repositorio, como .venv, .env y archivos generados por herramientas como Jupyter o VS Code.

Ver qué detectó Git:

```python
git status
```

Podrías ver:

Untracked files:
- .gitignore
- README.md
- requirements.txt
- 00_Configuracion/
- 01_Limpieza_de_datos/

Y probablemente .venv no aparecerá, porque está en .gitignore.

## El concepto más importante: commit

Git trabaja principalmente con:

```python
Archivos
   ↓
git add
   ↓
Staging
   ↓
git commit
   ↓
Historial
```

## Primer git add

Cuando estés listo para registrar tus archivos:

```python
git add .
```

El . significa: Agrega los cambios de la carpeta actual y sus subcarpetas, respetando .gitignore.

Después:

```python
git status
```

Ahora Git mostrará los archivos preparados para el commit.

## Crear tu primer commit

Luego de agregar los archivos a guardar con add . y verificar con status procedemos a ejecutar el commit

```python
git commit -m "Configuración inicial del proyecto"
```

## Ahora viene GitHub

Hasta el momento Todavía no existe conexión con GitHub. Ahora debemos crear el repositorio remoto.

- En GitHub creas un repositorio llamado: **Data-Analysis**

## Conectar tu proyecto local con GitHub

GitHub te proporcionará una dirección del repositorio algo asi:

```python
https://github.com/TU-USUARIO/Data-Analysis.git
```

Ahora En PowerShell:

```python
git remote add origin https://github.com/TU-USUARIO/Data-Analysis.git
```

Ahora lo comprobamos con:

```python
git remote -v
```

Deberíamos obtener algo similar a:

```python
origin  https://github.com/TU-USUARIO/Data-Analysis.git (fetch)
origin  https://github.com/TU-USUARIO/Data-Analysis.git (push)
```

Primero verifica la rama:

```python
git branch
```

Deberia aparecer como *master o main, usualmente aparece master pero para mayor facilidad lo vamos a pasar a main ya que es mas comun usarlo 

```python
git branch -M main
```

Por ultimo como ya antes habiamos inciado el **add .** y el commit procedemos a realizar el git push 

```python
git push -u origin main
```

Este proceso permite subir automáticamente todos los cambios realizados en el proyecto, incluyendo los archivos que han sido confirmados mediante los commits efectuados previamente en el repositorio.

## Flujo normal a trabajar

# Commit para el proceso a trabajar 

Algunas veces es necesario realizar un commit únicamente sobre un archivo o una carpeta específica en la que se estuvo trabajando. En lugar de utilizar git add ., que agrega todos los cambios del directorio actual, se puede utilizar git add <nombre_archivo> o git add <nombre_carpeta> para seleccionar únicamente los elementos que se desean incluir en el commit. Esto permite tener un mejor control sobre los cambios que se registran en el repositorio.

Agregar un archivo específico:

```python
git add 1. Introduccion/1.1 Introduccion.ipynb
```

Agregar una carpeta específica:

```python
git add 1. Introduccion/
```
