## Creación del archivo requeriments.txt

Después de instalar las librerías necesarias para el análisis de datos, se debe crear un archivo llamado requirements.txt. Este archivo permite registrar las versiones específicas de cada librería y componente utilizado en el proyecto, garantizando que el entorno pueda reproducirse de manera consistente y evitando posibles conflictos de compatibilidad en el futuro.

Después de instalar tus librerías con el entorno activado:

```python
pip freeze > requirements.txt
```

¿Por qué es importante?

Porque si dentro de seis meses pierdes ese entorno, no tienes que recordar qué instalaste.

Simplemente:

```python
python -m venv .venv
.\.venv\Scripts\Activate.ps1
pip install -r requirements.txt
```

De esta manera, el entorno puede reconstruirse fácilmente utilizando el archivo requirements.txt, instalando automáticamente todas las librerías utilizadas en el proyecto junto con las versiones especificadas. Esto permite mantener un entorno de trabajo consistente y reproducible.
